# Peek at `Type1Diabetes.parquet`

Quick look at the schema and first rows using DuckDB. The file is ~1.3 GB / ~133M rows, so we query it in place instead of loading it all into pandas.

```python
import duckdb
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 200)

PARQUET = "../data/Type1Diabetes.parquet"
con = duckdb.connect()
```

## 1. Schema (`DESCRIBE`)

```python
schema_df = con.sql(f"DESCRIBE SELECT * FROM '{PARQUET}'").df()
schema_df
```

## 2. All column names

```python
columns = schema_df["column_name"].tolist()
print(f"{len(columns)} columns:")
for name, dtype in zip(schema_df["column_name"], schema_df["column_type"]):
    print(f"  {name:<32} {dtype}")
```

## 3. First 5 rows

```python
head_df = con.sql(f"SELECT * FROM '{PARQUET}' LIMIT 5").df()
head_df
```
